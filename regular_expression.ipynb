```python
import re


def check_punctuation(text):
    result = re.search(r"[!?.]", text)
    return result != None


print(check_punctuation("This is a sentence that ends with a period."))  # True
print(check_punctuation("This is a sentence fragment without a period"))  # False
print(check_punctuation("Aren't regular expressions awesome?"))  # True
print(check_punctuation("Wow! We're really picking up some steam now!"))  # True
print(check_punctuation("End of the line"))  # False
```

### More about regular expression (Repeated Matches)

```python
import re
print(re.search(r"Py.*n", "Pygmalion"))
# .* combination can match any words
print(re.search(r"Py.*n", "Python Programming"))
print(re.search(r"Py[a-z]*n", "Python Programming"))
print(re.search(r"Py[a-z]*n", "Pyn"))
```

```python
import re
# + character matches 1 or more currences
print(re.search(r"o+l+", "goldfish"))
print(re.search(r"o+l+", "woolly"))
print(re.search(r"o+l+", "boil"))
```

```python
import re
print(re.search(r"p?each", "To each their own"))
# ? means optional
print(re.search(r"p?each", "I like peaches"))
```

```python
import re


def repeating_letter_a(text):
    result = re.search(r"a.*a", text, re.IGNORECASE)
    return result != None


print(repeating_letter_a("banana"))         # True
print(repeating_letter_a("pineapple"))      # False
print(repeating_letter_a("Animal Kingdom"))  # True
print(repeating_letter_a("A is for apple"))  # True
```

```python
import re
# now looks strictly for a literal period followed by "com"
print(re.search(r".com", "welcome"))
print(re.search(r"\.com", "welcome"))
print(re.search(r"\.com", "mydomain.com"))
```

```python
import re
# The \w matches any "word character" (letters, numbers, and the underscore _).
# The asterisk * means "match this zero or more times consecutively."
print(re.search(r"\w*", "This is an example"))
print(re.search(r"\w*", "And_this_is_another"))
```

```python
import re
"""
^ Start and $ end
"""
print(re.search(r"A.*a", "Argentina"))
print(re.search(r"A.*a", "Azerbaijan"))
print(re.search(r"^A.*a$", "Australia"))
```

### Capturing groups

```python
import re
# \w* will match letters, words, numbers and underscore
result = re.search(r"^(\w*), (\w*)$", "Lovelace, Ada")
print(result)
print(result.groups())
print(result[0])
print(result[1])
print(result[2])
"{} {}".format(result[2], result[1])
```

```python
import re


def rearrange_name(name):
    result = re.search(r"^(\w*), (\w*)$", name)
    if result is None:
        return name
    return "{} {}".format(result[2], result[1])


rearrange_name("Lovelace, Ada")
```

```python
import re


def rearrange_name(name):
    result = re.search(r"^([\w \.-]*), ([\w \.-]*)$", name)
    if result == None:
        return name
    return "{} {}".format(result[2], result[1])


rearrange_name("Hopper-mam, Grace M.")
```

```python
import re


def rearrange_name(name):
    result = re.search(r"^(\w*), (\w.*)$", name)
    # . is matches any kind of character
    if result == None:
        return name
    return "{} {}".format(result[2], result[1])


name = rearrange_name("Kennedy, John F.")
print(name)
```

```python
def find_url(website):
    pattern = (r"\w{5}://\w*\w*\.\w*\.\w*")  # enter the regex pattern here
    result = re.findall(pattern, website)  # enter the re method here
    return result


# Should return ['https://www.coursera.com', 'https://www.python.org']
print(find_url("Go to the website https://www.coursera.com find more information about Google Certificate Programs. Then, visit https://www.python.org/ to learn more about Python. "))
# Should return ['https://www.google.com']
print(find_url("You can find anything on https://www.google.com!"))
# Should return []
print(find_url("You can find anything on http://www.google.com!"))
print(find_url("Check out python.org!"))  # Should return []
```

```python
import re


def parse_city_state(text):
    pattern = r"^([^,.]+)[,.]\s+(.*)$"
    result = re.search(pattern, text)
    if result is None:
        return ""
    return result.group(2)


print(parse_city_state("Hamilton, MN"))             # MN
print(parse_city_state("Albuquerque, New Mexico"))  # New Mexico
print(parse_city_state("Portland. Oregon"))         # Oregon
```

```python
import re


def find_eid(report):
    # \d means any digit
    pattern = r"\b[BC]-\d{7,8}\b"  # enter the regex pattern here
    result = re.findall(pattern, report)  # enter the re method here
    return result


print(find_eid("Employees B-1234567 and C-12345678 worked with products X-123456 and Z-123456789"))
# Should return ['B-1234567', 'C-12345678']

print(find_eid("Employees B-1234567 and C-12345678, not employees b-1234567 and c-12345678"))
# Should return ['B-1234567', 'C-12345678']
```

```python
def replace_domain(address, old_domain, new_domain):
    old_domain_pattern = r'' + old_domain + '$'
    address = re.sub(old_domain_pattern, new_domain, address)
    return address
```
